```python
# Libraries were gonna need
import pandas as pd
```

```python
# Our Data, feel free to change if you want
# Leave signifigance Data as all "Unsure"
# Since we dont really know if it is just yet.

data = {
    "Tests" : [0.001, 0.008, 0.015, 0.03, 0.04, 0.06, 0.12, 0.45] ,
    "Significance_Before" : ["Unsure", "Unsure", "Unsure", "Unsure", "Unsure", "Unsure", "Unsure", "Unsure"]
}
q = 0.05
df = pd.DataFrame(data)
df.sort_values(by="Tests", ascending=True, inplace=True)
m = len(df)
print(df)
```

```python
# Show and make the table before modifications
# We will also order from ascending order, since BH works best in such way
df = pd.DataFrame(data)
df.sort_values(by="Tests", ascending=True)
print(df)
```

```python
# Iterate through "Tests", use formula to test signifigance
# If it satisfies the formula, add "Yes" to the new signifigance column
# Otherwise keep no, clean up all null/NaN values

k = 1

for value in df["Tests"]:
  if value <= (k/m) * q:
    df.loc[df['Tests'] == value, "Signifigance_After"] = "Yes"
  else:
    df.loc[df['Tests'] == value, "Signifigance_After"] = "No"
  k += 1

df.fillna("No", inplace=True)
print(df)
```
